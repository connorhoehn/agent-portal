## Conventions

- **Config sharing:** `config.json` loaded at top of every notebook, saved at bottom when new resources created
- **Resource tracking:** Every resource creation calls `tracker.add_*()` from `cleanup_utils.py`
- **Progress feedback:** `print("\u2713 ...")` for success, `print("\u231b ...")` for waiting
- **Idempotent:** Check if resource exists before creating
- **openCypher:** Neptune Analytics uses openCypher ONLY (no Gremlin/SPARQL)
- **MERGE over CREATE:** Idempotent graph writes
- **Titan Embeddings V2:** 1024 dimensions everywhere

````python
# Quick validation: check config.json exists
import json
from pathlib import Path

config_path = Path("config.json")
if config_path.exists():
    config = json.loads(config_path.read_text())
    print("\u2713 config.json found")
    print(f"  Region: {config['REGION']}")
    print(f"  Project prefix: {config['PROJECT_PREFIX']}")
    populated = {k: v for k, v in config.items() if v is not None and v != {}}
    print(f"  Populated fields: {len(populated)}/{len(config)}")
else:
    print("\u2717 config.json not found. Run 01-setup-credentials.ipynb first.")
````
