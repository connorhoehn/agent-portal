# OrgIQ — AWS AgentCore Accelerator

## Prerequisites

- [ ] AWS SSO login active (`aws sso login --profile bedrock`)

- [ ] AgentCore CLI installed (`npm i -g @aws/agentcore`)
- [ ] Region set to `us-east-1`

## Architecture

```
User → Supervisor Agent (AG-UI) → Search Agent (A2A)
                                 → Scoring Agent (A2A)
                                 → Advisor Agent (A2A)

Ingestion Agent → Graph Builder Agent → Neptune Analytics
                                      → Knowledge Base (OpenSearch)
```
